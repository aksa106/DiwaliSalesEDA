## dropna only removes values with nan 
    ommon Parameters
    Code	Meaning
    df.dropna()	Remove rows with NaN.
    df.dropna(axis=1)	Remove columns with NaN.
    df.dropna(inplace=True)	Modify the original DataFrame.
    df.dropna(subset=["Amount"])	Remove only rows where Amount is NaN.


    Function	Purpose
    drop(columns=["Status"])	Remove specific column(s) you choose.
    dropna(axis=1)	Remove every column that contains at least one NaN (or all NaN, depending on parameters).


    drop() → I decide what to remove.
    dropna() → Pandas decides based on missing values. 


    Rule of Thumb
    Known rows? → drop(index=...)
    Known columns? → drop(columns=...)
    Rows with NULLs? → dropna()
    Replace NULLs? → fillna()

    Duplicates — Summary
    1. Count Duplicate Rows
    df.duplicated().sum()

    Counts duplicate rows.

    2. Show Duplicate Rows
    df[df.duplicated()]

    Shows only the duplicate copies (keeps the first occurrence).

    3. Show All Rows Involved ⭐
    df[df.duplicated(keep=False)]

    Shows both the original and duplicate rows.

    4. Remove Duplicate Rows
    df.drop_duplicates(inplace=True)

    Removes duplicate rows.

    5. Check Duplicates in a Specific Column
    df["User_ID"].duplicated().sum()

    Counts duplicate values in User_ID.

    6. Show Duplicate Values of a Column
    df[df["User_ID"].duplicated(keep=False)]

    Shows all rows where User_ID is duplicated.

    7. Remove Duplicates Based on a Column
    df.drop_duplicates(subset=["User_ID"], inplace=True)

    Keeps only one row per User_ID.

    ⚠️ Use only if duplicates in that column are errors (e.g., Employee_ID). Do not use it if repeated values are expected (e.g., customers making multiple purchases).

    Rule of Thumb
    duplicated() → Find duplicates.
    duplicated(keep=False) → Investigate duplicates.
    drop_duplicates() → Remove duplicates.
    subset=[...] → Check/remove duplicates using specific column(s).
